```python
class Coche:
    def __init__(self, marca): 
        self.marca = marca
        self.velocidad = 0
    def acelerar(self, inc): 
        self.velocidad += inc
    # 3. def frenar(self, dec): resta dec a self.velocidad; si queda por
    #    debajo de 0, vuelve a ponerla a 0.
    def frenar(self, dec):
        self.velocidad -= dec
        if self.velocidad < 0:
            self.velocidad = 0
    
a = Coche("Seat")
b = Coche("Tesla")
a.acelerar(50)
b.acelerar(120)
b.frenar(200)
print(f"{a.marca} va a {a.velocidad} km/h")
print(f"{b.marca} va a {b.velocidad} km/h")

# 4. Fuera de la clase: crea a = Coche("Seat") y b = Coche("Tesla").
# 5. Llama a a.acelerar(50), b.acelerar(120) y b.frenar(200).
# 6. Imprime la marca y la velocidad de cada uno.
```

El mismo gestor de biblioteca del repositorio, pero con dos clases que colaboran:

1. Libro(titulo, autor), con un método datos() que devuelva "titulo - autor";
2. Biblioteca(), que arranque con self.libros = [] y tenga agregar(libro), eliminar(titulo), buscar(titulo) (devuelve el libro o None) y mostrar().
Crea una biblioteca, agrega dos libros (por ejemplo "1984" de Orwell y "Un mundo feliz" de Huxley), busca uno por su título, muéstralos todos y elimina uno.

Cómo abordarlo, paso a paso:

**Empieza por Libro.** Su __init__ guarda el título y el autor en self, y datos() los devuelve juntos con return (si los imprimes, el método devuelve None). Pruébala sola antes de seguir: crea un libro e imprime lo que devuelve su datos().

**Después, Biblioteca.** Guarda los libros en una lista, self.libros, que nace vacía en su __init__. Los cuatro métodos trabajan sobre esa misma lista, y lo que hay dentro son objetos Libro enteros: para comparar un título miras libro.titulo, y para imprimir uno, libro.datos().

Cada método es algo que ya sabes hacer con una lista:** agregar añade un elemento, buscar la recorre con un for hasta dar con el título, eliminar quita uno y mostrar la recorre imprimiendo. Si buscar no encuentra el título, devuelve None: así quien la llama puede preguntar if resultado is None: y saber que no está.

**Por último, úsala** con los dos libros del enunciado. Ve método a método y pulsa Ejecutar después de cada uno: es más fácil encontrar un fallo cuando solo has cambiado una cosa.
Si te atascas, pulsa Pista: cada vez que la pulses te dará un poco más.

```python
class Libro:
    def __init__(self, titulo, autor):  
        self.titulo = titulo
        self.autor = autor
    def datos(self):
        return f"{self.titulo} de {self.autor}"
class Biblioteca:
    def __init__(self):
        self.libros = []
    def agregar(self, libro):
        self.libros.append(libro)
    def buscar(self, titulo):
        for libro in self.libros:
            if libro.titulo == titulo:
                return libro
        return None
    def eliminar(self,titulo):
        libro_encontrado = self.buscar(titulo)
        if libro_encontrado:
            self.libros.remove(libro_encontrado)
    def mostrar(self):
        for libro in self.libros:            
            print(libro.datos())
            
mi_biblioteca = Biblioteca()
libro1 = Libro("1984", "George Orwell")
libro2 = Libro("Cien años de soledad", "Gabriel García Márquez")

mi_biblioteca.agregar(libro1)
mi_biblioteca.agregar(libro2)

resultado = mi_biblioteca.buscar("1984")
if resultado is not None:
    print("Libro encontrado:", resultado.datos())
else:
    print("Libro no encontrado")
    
print("\nLista de libros en la biblioteca:")
mi_biblioteca.mostrar()
mi_biblioteca.eliminar("1984")
print("\nLista de libros después de eliminar '1984':")
mi_biblioteca.mostrar()
```
