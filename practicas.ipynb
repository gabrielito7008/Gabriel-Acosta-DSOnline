**Qué hay que conseguir:** una función que recibe un precio y un descuento en 
tanto por ciento y te devuelve el precio ya rebajado.

precio_final(100, 15) devuelve 85.0: a 100 se le quita su 15 %, que son 15.
precio_final(100) devuelve 100: si no le pasas descuento, vale 0 y el precio no cambia.
precio_final(base=80, descuento=25) devuelve 60.0: los mismos datos, pasados por su nombre.
La cuenta es la de siempre: el descuento en euros es base * descuento / 100, y el precio final es la base menos ese descuento. 
La consola trae la primera línea de la función escrita: tú escribes lo de dentro y las tres pruebas.

```python
def precio_final(base, descuento=0):
    resultado = base - (base * descuento / 100)
    return round(resultado, 2)

    # base es el precio de partida y descuento el tanto por ciento que se quita
    # (un 15 quiere decir un 15 %).
    # descuento=0 quiere decir: si al llamar a la función no se dice nada, vale 0.
    # Aquí dentro calcula el precio rebajado y entrégalo con return,
    # redondeado a 2 decimales con round(numero, 2).
    # pass no hace nada: solo está para que la función no quede vacía. Bórral

# Fuera de la función, pruébala imprimiendo lo que devuelve en tres casos:
print(precio_final(100))                    # -> tiene que salir 100
print(precio_final(100, 15))                # -> tiene que salir 85.0
print(precio_final(base=80, descuento=25))  # -> tiene que salir 60.0
```

Qué hay que conseguir: la misma idea que la hucha de arriba, con un coche. La clase Coche guarda su marca y su velocidad, que empieza en 0, y tiene un método acelerar que sube la velocidad y te devuelve un texto contándolo.

Al crear mi_coche = Coche("Seat") y acelerarlo dos veces, la salida tiene que parecerse a esta:

1. Seat va a 30 km/h
2. Seat va a 50 km/h
3. Seat 50


```python
class Coche:
    def __init__(self, marca):
        self.marca = marca
        self.velocidad = 0

    def acelerar(self, incremento):
        self.velocidad += incremento
        return f"{self.marca} va a {self.velocidad} km/h"
    
mi_coche = Coche("Seat")
print(mi_coche.acelerar(30))
print(mi_coche.acelerar(20))
print(mi_coche.marca)
print(mi_coche.velocidad)

# 3. Fuera de la clase: crea mi_coche = Coche("Seat"),
#    imprime lo que devuelve mi_coche.acelerar(30) y luego mi_coche.acelerar(20),
#    y al final imprime mi_coche.marca y mi_coche.velocidad.
```

Qué hay que conseguir: comprobar que dos coches creados con la misma clase llevan cada uno su propia velocidad, y que frenar nunca deja un coche con velocidad negativa.

1. Escribe otra vez la clase Coche, con marca, velocidad que empieza en 0 y acelerar(self, inc), que suma inc a la velocidad (aquí no hace falta que devuelva nada).
2. Añádele frenar(self, dec), que resta dec a la velocidad. Si al restar quedaría por debajo de 0, la velocidad se queda en 0, como hace sacar en la hucha de arriba: un coche que va a 120 y frena 200 se para, no va a -80.
3. Crea dos coches, a = Coche("Seat") y b = Coche("Tesla"), y haz a.acelerar(50), b.acelerar(120) y b.frenar(200).
4. Imprime la marca y la velocidad de cada uno.

La salida tiene que parecerse a esta: frenar b no ha tocado a a.

Seat: 50 km/h
Tesla: 0 km/h

```python
class Coche:
    def __init__(self, marca): 
        self.marca = marca
        self.velocidad = 0
    def acelerar(self, inc): 
        self.velocidad += inc
    # 3. def frenar(self, dec): resta dec a self.velocidad; si queda por
    #    debajo de 0, vuelve a ponerla a 0.
    def frenar(self, dec):
        self.velocidad -= dec
        if self.velocidad < 0:
            self.velocidad = 0
    
a = Coche("Seat")
b = Coche("Tesla")
a.acelerar(50)
b.acelerar(120)
b.frenar(200)
print(f"{a.marca} va a {a.velocidad} km/h")
print(f"{b.marca} va a {b.velocidad} km/h")

# 4. Fuera de la clase: crea a = Coche("Seat") y b = Coche("Tesla").
# 5. Llama a a.acelerar(50), b.acelerar(120) y b.frenar(200).
# 6. Imprime la marca y la velocidad de cada uno.
```
